# ExpenseFlow ML Service - Anomaly Detection Demo

This notebook demonstrates how to interact with the machine learning microservice (FastAPI on Port 8000) for anomaly scoring and receipt extraction.

```python
import requests
import json

ML_SERVICE_URL = 'http://localhost:8000'

# Check health
res = requests.get(f'{ML_SERVICE_URL}/health')
print("ML Service Status:", res.json())
```

## 1. Anomaly Detection

Submit an expense payload to see the anomaly flag and fraud probability score.

```python
payload = {
    "amount": 150000,
    "category": "Travel",
    "description": "Weekend trip to Bali",
    "user_role": "employee"
}

res = requests.post(f'{ML_SERVICE_URL}/api/anomaly/detect', json=payload)
print(json.dumps(res.json(), indent=2))
```
